4. Dataset Description

"For this project, a sample dataset of five beauty and wellness products is used. It contains product names, prices, customer ratings and sales figures. This dataset is created for learning and demonstration purposes and does not represent actual company sales data."

```python
data = {
   "Product": ["Face wash", "sunscreen", "face serum",
                "moisturizer", "shampoo"],

   "Price": [250, 450, 600, 350, 300],

  "Rating": [4.2, 4.5, 4.7, 4.1, 4.0],

  "Sales": [120, 180, 200, 100, 90]
}

df = pd.DataFrame(data)

print(df)
```

5. Sales Analysis

"This section analyses the sales of different beauty and wellness products to identify which product performs best. It also calculates the average sales to understand the overall sales performance."

```python
highest_sales = df["Sales"].max()

best_product = df[df["Sales"] == highest_sales]

print("Highest Selling Product:")
print(best_product)
```

6. Product Sales Visualization

"This bar chart compares the sales of different beauty and wellness products. It helps us understand which products have higher and lower sales."

```python
print("Average Sales:", df["Sales"].mean())
print("Average Rating:", df["Rating"].mean())
```

```python
import matplotlib.pyplot as plt

plt.bar(df["Product"], df["Sales"])

plt.title("Beauty Products Sales Analysis")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

7. Customer Rating Analysis

"This section analyses customer ratings of different beauty products. It helps identify which products receive better customer feedback."

```python
plt.bar(df["Product"], df["Rating"])

plt.title("Customer Rating Analysis")
plt.xlabel("Beauty Products")
plt.ylabel("Customer Rating")

plt.ylim(0, 5)
plt.show()
```

```python
highest_rating = df["Rating"].max()

print("Highest Customer Rating:", highest_rating)

print(df[df["Rating"] == highest_rating])
```

8. Monthly Sales Trend Analysis

"This section studies how sales change over time. A line graph is used to understand the monthly sales pattern and identify whether sales are increasing or decreasing."

```python
monthly_data ={
    "Month": ["January", "February", "March",
              "April", "May", "June"],

    "Sales": [100, 120, 150, 140, 180, 220]
}

monthly_df = pd.DataFrame(monthly_data)

print(monthly_df)
```

```python
plt.plot(monthly_df["Month"], monthly_df["Sales"],
        marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid()
plt.show()
```

9. Future Sales Prediction

"Linear Regression is used to study the relationship between time and sales. Based on the sample monthly sales data, the model estimates sales for the next two months. These predictions are only for demonstration and are not actual market forecasts."

```python
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4], [5], [6]]
y = monthly_df["Sales"]

model = LinearRegression()
model.fit(X, y)

new_month = [[7], [8]]
predictions = model.predict(new_month)

print("Future Sales Predictions:")

print("July Sales:", round(predictions[0], 2))
print("August Sales:", round(predictions[1], 2))
```

10. Actual and Predicted Sales Visualization

"This graph shows the actual sales from January to June and the estimated sales for July and August. It provides a visual comparison of past sales and future predictions."

```python
import matplotlib.pyplot as plt

months = ["January", "February", "March",
          "April", "May", "June", "July", "August"]

all_sales = [100, 120, 150, 140, 180, 220,
             predictions[0], predictions[1]]

plt.figure(figsize=(9, 5))

plt.plot(months, all_sales, marker="o")

plt.title("Actual and Predicted Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)
plt.show()
```
